```python
import numpy as np
```

```python
temperatures = np.array([45, 48, 51, 49, 53, 68, 74, 47, 82, 50])
low, high = 40, 70
```

```python
in_range = (temperatures >= low) & (temperatures <= high)
outliers = temperatures[~in_range]
```

```python
print("Mean temperature:", temperatures.mean())
print("Median temperature:", np.median(temperatures))
print(f"Temperatures outside the normal range ({low}-{high}):", outliers)
print("Number of temperatures outside the normal range:", len(outliers))
```

This code snippet calculates the mean and median of a set of temperature readings, identifies which readings fall outside the normal range of 40 to 70 degrees, and counts how many readings are outside that range.
