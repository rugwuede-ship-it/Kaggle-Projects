```python
import numpy as np
```

```python
machines = np.array([
    [10, 12, 11, 13],
    [15, 14, 16, 15],
    [8,  9,  7, 10]
])
```

```python
machine_means = machines.mean(axis=1)
top_machine = machine_means.argmax() + 1  # +1 because machines are numbered from 1
```

```python
print("Mean energy consumption for each machine:", machine_means)
print("Overall mean energy consumption:", machines.mean())
print(f"Highest average consumption: Machine {top_machine} ({machine_means.max()})")
```
